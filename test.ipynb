```python
def build_comparison_lines(regional_df: pd.DataFrame,
                           national_df: pd.DataFrame) -> str:
    """
    Convert regional & national DataFrames (same cols) into a bullet-list string
    of the form
        **Indicator (unit)**
        2020: 12.3 Mbps – Veliko Gradište | 18.1 Mbps – National avg
        2021: ...
    """
    lines = []
    for col in regional_df.columns.drop(["GID_2","SERBIAN_NAME_CYRILLIC", "ENGLISH_NAME", "year"]):
        unit = ""  # ► fill if you store units in a lookup dict
        lines.append(f"**{col}{unit}**")
        for _, r_row in regional_df.iterrows():
            year = int(r_row["year"])
            region_val = r_row[col]
            nat_val   = national_df.loc[national_df["year"] == year, col].iloc[0]
            region_name = r_row["ENGLISH_NAME"]
            lines.append(
                f"{year}: {region_val:.2f} – {region_name} | {nat_val:.2f} – National avg"
            )
        lines.append("")   # blank line between indicators
    return "\n".join(lines)


regional = df[df['ENGLISH_NAME'] == 'Bor'].reset_index(drop=True)
national = df.groupby('year')[df.columns[4:]].mean().reset_index()

build_comparison_lines(regional, national)
```
