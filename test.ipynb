````python
def build_comparison_lines(regional_df: pd.DataFrame,
                           national_df: pd.DataFrame) -> str:
    """
    Convert regional & national DataFrames (same cols) into a bullet-list string
    of the form
        **Indicator (unit)**
        2020: 12.3 Mbps – Veliko Gradište | 18.1 Mbps – National avg
        2021: ...
    """
    lines = []
    for col in regional_df.columns.drop(["GID_2","SERBIAN_NAME_CYRILLIC", "ENGLISH_NAME", "year"]):
        unit = ""  # ► fill if you store units in a lookup dict
        lines.append(f"**{col}{unit}**")
        for _, r_row in regional_df.iterrows():
            year = int(r_row["year"])
            region_val = r_row[col]
            nat_val   = national_df.loc[national_df["year"] == year, col].iloc[0]
            region_name = r_row["ENGLISH_NAME"]
            lines.append(
                f"{year}: {region_val:.2f} – {region_name} | {nat_val:.2f} – National avg"
            )
        lines.append("")   # blank line between indicators
    return "\n".join(lines)


regional = df[df['ENGLISH_NAME'] == 'Bor'].reset_index(drop=True)
national = df.groupby('year')[df.columns[4:]].mean().reset_index()

build_comparison_lines(regional, national)
````

````python
import pandas as pd

def classify_road(asset_type: str) -> str:
    if asset_type is None:
        return "Other / Unknown"

    at = asset_type.lower()

    if at in ["motorway", "motorway_link", "trunk", "trunk_link"]:
        return "Motorway & Trunk"

    if at in ["primary", "primary_link"]:
        return "Primary / Arterial"

    if at in ["secondary", "secondary_link", "tertiary", "tertiary_link"]:
        return "Secondary & Tertiary"

    if at in ["residential", "unclassified", "living_street"]:
        return "Local Streets"

    if at in ["service", "road"]:
        return "Service & Access"

    if at in ["track"]:
        return "Tracks & Agricultural"

    if at in ["footway", "path", "pedestrian", "steps", "cycleway", "bridleway", "corridor"]:
        return "Non-motorized"

    # fallback
    return "Other / Unknown"

# Assuming you already have roads_clipped["asset_type"]
roads_clipped["road_class"] = roads_clipped["asset_type"].apply(classify_road)
print(roads_clipped["asset_type"].value_counts())
print("\nBy road_class:")
print(roads_clipped["road_class"].value_counts())
````

````python
roads_clipped = roads_clipped.to_crs(epsg=3857)
roads_clipped = roads_clipped.drop_duplicates(subset=['osmid'])
roads_clipped['length'] = roads_clipped.geometry.length / 1000
roads_clipped.groupby(['road_class'])['length'].sum().reset_index()
````

````python
import plotly.graph_objects as go

# Basic style mapping for road_class
class_color = {
    "Motorway & Trunk": "red",
    "Primary / Arterial": "orange",
    "Secondary & Tertiary": "yellow",
    "Local Streets": "green",
    "Service & Access": "blue",
    "Tracks & Agricultural": "sienna",
    "Non-motorized": "purple",
    "Other / Unknown": "lightgray",
}

class_width = {
    "Motorway & Trunk": 4,
    "Primary / Arterial": 3.5,
    "Secondary & Tertiary": 3,
    "Local Streets": 2,
    "Service & Access": 1.8,
    "Tracks & Agricultural": 1.5,
    "Non-motorized": 1.2,
    "Other / Unknown": 1.2,
}

fig = go.Figure()

for road_class, subset in roads_clipped.groupby("road_class"):
    lons, lats, texts = [], [], []

    for _, row in subset.iterrows():
        geom = row.geometry
        if geom is None:
            continue

        if geom.geom_type == "LineString":
            x, y = geom.xy
            lons.extend(list(x) + [None])
            lats.extend(list(y) + [None])
            texts.extend(
                [f"class: {road_class}<br>type: {row['asset_type']}<br>name: {row['name']}"]
                * (len(x) + 1)
            )

        elif geom.geom_type == "MultiLineString":
            for part in geom.geoms:
                x, y = part.xy
                lons.extend(list(x) + [None])
                lats.extend(list(y) + [None])
                texts.extend(
                    [f"class: {road_class}<br>type: {row['asset_type']}<br>name: {row['name']}"]
                    * (len(x) + 1)
                )

    if not lons:
        continue

    fig.add_trace(
        go.Scattermapbox(
            lon=lons,
            lat=lats,
            mode="lines",
            line=dict(
                width=class_width.get(road_class, 1.5),
                color=class_color.get(road_class, "lightgray"),
            ),
            hoverinfo="text",
            text=texts,
            name=road_class,  # legend entry
        )
    )

# Municipal boundary
boundary = muni_geom.boundary
bx, by = boundary.coords.xy
fig.add_trace(
    go.Scattermapbox(
        lon=list(bx),
        lat=list(by),
        mode="lines",
        line=dict(width=3, color="black"),
        hoverinfo="skip",
        showlegend=False,
    )
)

# Center map
muni_centroid = muni_geom.centroid
center_lon, center_lat = float(muni_centroid.x), float(muni_centroid.y)

fig.update_layout(
    margin={"l": 0, "r": 0, "t": 0, "b": 0},
    mapbox_style="open-street-map",
    mapbox_zoom=11,
    mapbox_center={"lon": center_lon, "lat": center_lat},
    legend=dict(title="Road class"),
)

fig.show()
````
